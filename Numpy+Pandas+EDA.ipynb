import numpy as np

```
#creating arrays
a= np.array([10, 20, 30, 40,50])
```

```
print(a.shape) #shape
print(a.ndim) #dimension
```

```
print(a[1]) #indexing
print(a[1:4]) #Slicing
```

```
#Boolean masking
print(a[a>25])
```

```
#math Operations
print(np.mean(a)) #mean
print(np.sum(a)) #sum
print(np.std(a)) #standard deviation
```

```
#reshaping
b = a.reshape(5,1)
print(b)
```

```
print(a*2) #Vectorized operations
#Vectorized operations are a way of performing computations on entire arrays (or vectors) of data at once, instead of using explicit loops to process one element at a time
```

```
#missing value
c = np.array([1,2, np.nan, 4])
print(np.nanmean(c))
print(c)
c1 = np.nanmean(c)
print(c1)
c2 = np.append(c, c1)
print(c2)
```

NumPy uses contiguous memory, C-based backend, and vectorized operations, which reduce Python-level loops and overhead, making it much faster

Vectorization means performing operations on entire arrays at once without using explicit loops, which improves speed and performance.

#Pandas + EDA

Pandas is a python library used for data manipulation and analysis it provides DataFrame and seies structures that allows us to clean filter group , merge, and analyze structured data efficiency

```
import pandas as pd

#create DataFrame
data = {
    "name": ["a","b","c","d"],
    "age": [10,20,None,40],
    "salary": [3000, 4000,5000,None]
}
```

```
df = pd.DataFrame(data)
print(df)
```

```
#checking the missing value
print(df.isnull())
```

```
#Filling the missing value
df["age"].fillna(df["age"].mean(), inplace=True)
df["salary"].fillna(df["salary"].mean(), inplace=True)

print(df.isnull())

```

```
#drop rows with missing values
#df.dropna()

#filtering
print(df[df["age"]>24])
```

```
#sorting
print(df.sort_values(by="salary"))
```

```
#Grouping
print(df.groupby("age").mean(numeric_only=True))
```

```
#Exploratory Data Analysis(EDA)
#EDA is the process of exploring data to understand its structure, patterns, trends, and anomalies before applying any modeling or decision-making. It helps in identifying missing values, outliers, correlations, and overall data behavior.

import pandas as pd
data = {
    "product": ["A","B","C","D","E"],
    "sales":[200, 300 , None , 400 , 500],
    "Profit":[20,30,25,None,50]
}

df = pd.DataFrame(data)
```

```
print(df)
```

```
print(df.isnull())
```

```
df["sales"].fillna(df["sales"].mean(),inplace=True)
df["Profit"].fillna(df["Profit"].mean(), inplace = True)
print(df)
```

```
print(df[df["sales"]>300])
```
